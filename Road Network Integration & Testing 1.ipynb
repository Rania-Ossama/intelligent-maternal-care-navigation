```python
import os
import osmnx as ox

ox.settings.overpass_url = "https://overpass.kumi.systems/api"

save_dir = "upper_egypt_road_graphs"
os.makedirs(save_dir, exist_ok=True)

governorates = {
    "Beni_Suef": "Beni Suef Governorate, Egypt",
    "Minya": "Minya Governorate, Egypt",
    "Assiut": "Assiut Governorate, Egypt",
    "Sohag": "Sohag Governorate, Egypt",
    "Qena": "Qena Governorate, Egypt",
    "Luxor": "Luxor Governorate, Egypt",
    "Aswan": "Aswan Governorate, Egypt"
}

downloaded_graphs = {}

for name, place in governorates.items():
    filepath = os.path.join(save_dir, f"{name}.graphml")

    if os.path.exists(filepath):
        print(f"\n{name}: loading saved graph...")
        G_part = ox.load_graphml(filepath)
    else:
        print(f"\n{name}: downloading...")
        try:
            G_part = ox.graph_from_place(
                place,
                network_type="drive",
                simplify=True
            )
            ox.save_graphml(G_part, filepath)
            print(f"{name}: downloaded and saved.")
        except Exception as e:
            print(f"{name}: FAILED — {type(e).__name__}: {e}")
            continue

    downloaded_graphs[name] = G_part
    print(f"Nodes: {len(G_part.nodes):,}")
    print(f"Edges: {len(G_part.edges):,}")

print("\nSuccessfully downloaded:", list(downloaded_graphs.keys()))
print("Governorates completed:", len(downloaded_graphs), "out of", len(governorates))
```

```python
import pandas as pd

pairs_to_check = [
    ("Beni_Suef", "Minya"),
    ("Minya", "Assiut"),
    ("Assiut", "Sohag"),
    ("Sohag", "Qena"),
    ("Qena", "Luxor"),
    ("Qena", "Aswan"),
]

for gov_a, gov_b in pairs_to_check:
    a = pd.read_parquet(
        OUTPUT_DIR / f"{gov_a}_nodes.parquet",
        columns=["id", "lon", "lat"]
    )
    b = pd.read_parquet(
        OUTPUT_DIR / f"{gov_b}_nodes.parquet",
        columns=["id", "lon", "lat"]
    )

    shared = a.merge(b, on="id", suffixes=("_a", "_b"))

    if shared.empty:
        print(f"{gov_a} - {gov_b}: no shared nodes")
        continue

    same_coords = (
        (shared["lon_a"] - shared["lon_b"]).abs().lt(1e-7)
        & (shared["lat_a"] - shared["lat_b"]).abs().lt(1e-7)
    )

    print(
        f"{gov_a} - {gov_b}: "
        f"{len(shared):,} shared IDs, "
        f"{same_coords.sum():,} matching coordinates, "
        f"{(~same_coords).sum():,} mismatches"
    )

    del a, b, shared
```

```python

import numpy as np
from scipy.spatial import cKDTree

# Convert longitude/latitude to 3D Earth-sphere coordinates.
# This lets us estimate geographic distance in kilometers.
def to_xyz(lon, lat):
    lon = np.radians(np.asarray(lon, dtype=float))
    lat = np.radians(np.asarray(lat, dtype=float))

    return np.column_stack([
        np.cos(lat) * np.cos(lon),
        np.cos(lat) * np.sin(lon),
        np.sin(lat)
    ])

# Build the main component's spatial index
main_ids = components[0]

main_nodes = nodes[nodes["id"].isin(main_ids)]
main_xyz = to_xyz(main_nodes["lon"], main_nodes["lat"])

tree = cKDTree(main_xyz)

results = []

for rank, component_ids in enumerate(components[1:15], start=2):
    part = nodes[nodes["id"].isin(component_ids)]

    xyz = to_xyz(part["lon"], part["lat"])
    distances, _ = tree.query(xyz, k=1)

    nearest_km = distances.min() * 6371

    results.append({
        "component": rank,
        "nodes": len(part),
        "center_lon": part["lon"].median(),
        "center_lat": part["lat"].median(),
        "nearest_main_component_km": nearest_km
    })

result_df = pd.DataFrame(results)

print(result_df.round(3).to_string(index=False))
```

```python
import numpy as np
import re

edges_path = OUTPUT_DIR / "Upper_Egypt_edges_deduplicated.parquet"

edges = pd.read_parquet(edges_path)

# Provisional fallback speeds for initial algorithm testing only.
# These are assumptions, not measured Egyptian road speeds.
default_speed = {
    "motorway": 100,
    "motorway_link": 60,
    "trunk": 80,
    "trunk_link": 50,
    "primary": 60,
    "primary_link": 40,
    "secondary": 50,
    "secondary_link": 35,
    "tertiary": 40,
    "tertiary_link": 30,
    "unclassified": 30,
    "residential": 30,
    "living_street": 10,
    "service": 15,
    "track": 15,
    "road": 30,
}

def parse_speed(value):
    if pd.isna(value):
        return np.nan

    match = re.search(r"\d+(?:\.\d+)?", str(value))
    if not match:
        return np.nan

    speed = float(match.group())

    if "mph" in str(value).lower():
        speed *= 1.609344

    return speed if speed > 0 else np.nan

parsed_speed = edges["maxspeed"].apply(parse_speed)

fallback_speed = (
    edges["highway"].map(default_speed)
    .fillna(30.0)
)

edges["speed_kmh"] = parsed_speed.fillna(fallback_speed)

edges["speed_source"] = np.where(
    parsed_speed.notna(),
    "OSM maxspeed",
    "Provisional highway default"
)

# length is in meters; travel time is in seconds.
edges["travel_time_s"] = (
    edges["length"] / 1000
) / edges["speed_kmh"] * 3600

weighted_path = OUTPUT_DIR / "Upper_Egypt_edges_weighted.parquet"
edges.to_parquet(weighted_path, index=False)

print("Saved:", weighted_path)
print("\nSpeed source:")
print(edges["speed_source"].value_counts())

print("\nTravel time summary (seconds per edge):")
print(edges["travel_time_s"].describe())

print("\nInvalid weights:", (
    edges["travel_time_s"].isna().sum()
    + (~np.isfinite(edges["travel_time_s"])).sum()
    + (edges["travel_time_s"] <= 0).sum()
))
```

```python
from scipy.sparse import csr_matrix
from scipy.sparse.csgraph import dijkstra

# 1. Sort nodes so OSM IDs map to compact integer indices
nodes_sorted = nodes.sort_values("id").reset_index(drop=True)
node_ids = nodes_sorted["id"].to_numpy(dtype=np.int64)

u_ids = edges["u"].to_numpy(dtype=np.int64)
v_ids = edges["v"].to_numpy(dtype=np.int64)
weights = edges["travel_time_s"].to_numpy(dtype=np.float64)

u_idx = np.searchsorted(node_ids, u_ids)
v_idx = np.searchsorted(node_ids, v_ids)

# Validate that every edge endpoint exists
assert np.all(u_idx < len(node_ids))
assert np.all(v_idx < len(node_ids))
assert np.array_equal(node_ids[u_idx], u_ids)
assert np.array_equal(node_ids[v_idx], v_ids)

# 2. Build a memory-efficient directed sparse graph
G_sparse = csr_matrix(
    (weights, (u_idx, v_idx)),
    shape=(len(node_ids), len(node_ids))
)
G_sparse.sum_duplicates()
G_sparse.sort_indices()

print("Sparse graph created.")
print("Nodes:", f"{G_sparse.shape[0]:,}")
print("Directed connections:", f"{G_sparse.nnz:,}")

# 3. Select real nodes from the road network
source_id = int(u_ids[0])
target_id = int(v_ids[100])

source_idx = int(np.searchsorted(node_ids, source_id))
target_idx = int(np.searchsorted(node_ids, target_id))

print("\nSource OSM node:", source_id)
print("Target OSM node:", target_id)

# 4. Run baseline Dijkstra using estimated travel time
distances, predecessors = dijkstra(
    csgraph=G_sparse,
    directed=True,
    indices=source_idx,
    return_predecessors=True
)

# 5. Reconstruct the route
if np.isfinite(distances[target_idx]):
    path_indices = []
    current = target_idx

    while current != source_idx:
        path_indices.append(current)
        current = int(predecessors[current])

        if current < 0:
            raise RuntimeError("Could not reconstruct the route.")

    path_indices.append(source_idx)
    path_indices.reverse()

    route_ids = node_ids[path_indices]
    total_seconds = float(distances[target_idx])

    print("\nReal-network Dijkstra test passed.")
    print("Route nodes:", len(route_ids))
    print("Estimated travel time (seconds):", round(total_seconds, 2))
    print("Estimated travel time (minutes):", round(total_seconds / 60, 2))
    print("First 10 route node IDs:", route_ids[:10].tolist())
else:
    print("\nNo directed route found between these two nodes.")
    print("We'll select a reachable target for the next test.")
```
